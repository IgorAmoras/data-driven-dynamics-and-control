# Feedforward Neural Networks (FNN)

FNNs são a forma mais básica de aprendizado. Propostas em 1900 e bolinha, ela funcionam fazendo uma composição de funções de ativação. O princípio dela é usar de cadamas de rede capazes de geralizar a relação entrada-saída do sistema.
FNNs são tidas como aproximadoras universais. Para precisarem ser construídas, FNNs necessitam de ter acesso à todos os estados do sistema.
O princípio básico da construção baseaia-se em,
$$
\hat{x}_{i, k+1} = f_\theta(x_{i,k}, u_{i,k}) := FNN(x_{i,k}, u_{i,k})
$$
onde cada FNN consiste em,
$$
FNN(\zeta) := (l_L \circ l_{L-1} \circ ... \circ l_1)(\zeta)
$$
Onde cada camada l, começa com a saída de sua antecessora, temdo como valor inicial $\zeta$. Cada camada l é representada como,
$$
l_{i+1}(l_i,\theta_{i+1}) = \sigma (W_{î+1}l_i + b_{i+1}), \quad i = 0,...,L - 1
$$
onde $\sigma$ é um função de ativação não linear, W são pesos e $b$ são vieses a serem aprendidos, coletados então para $[\theta^T_i, ... , \theta^T_L]^T$. 

Ou seja, cada camada atua como um função afim que é ponderada com base em no parâmetro de entrada, treinada.
