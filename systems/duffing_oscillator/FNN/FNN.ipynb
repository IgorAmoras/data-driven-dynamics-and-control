# Feedforward Neural Networks (FNN)

FNNs são a forma mais básica de aprendizado. Propostas em 1900 e bolinha, ela funcionam fazendo uma composição de camadas, cada uma com uma transformação afim seguida de uma função de ativação. O princípio dela é usar de camadas de rede capazes de aproximar a relação entrada-saída do sistema.
FNNs são tidas como aproximadoras universais sob condições adequadas. Na formulação de identificação do mapa de transição de estados apresentada no artigo, necessitamos de ter acesso a todos os estados do sistema para o treinamento supervisionado.
O princípio básico da construção baseaia-se em,
$$
\hat{x}_{i, k+1} = f_\theta(x_{i,k}, u_{i,k}) := FNN(x_{i,k}, u_{i,k})
$$
onde cada FNN consiste em,
$$
FNN(\zeta) := (l_L \circ l_{L-1} \circ ... \circ l_1)(\zeta)
$$
Onde cada camada $l_i$ recebe a saída $z_{i-1}$ de sua antecessora, tendo como valor inicial $z_0=\zeta$. Cada camada é representada como,
$$
z_{i+1}=l_{i+1}(z_i,\theta_{i+1}) = \sigma (W_{i+1}z_i + b_{i+1}), \quad i = 0,...,L - 1
$$
onde $\sigma$ é uma função de ativação, em geral não linear, $W$ são pesos e $b$ são vieses a serem aprendidos, coletados então em $\theta=[\theta^T_1, ... , \theta^T_L]^T$. 
Para uma família de FNNs com ativação adequada e largura suficiente ser tida como aproximadora universal de funções contínuas, estabelecemos formalmente que,
$$
\forall \varepsilon > 0, \quad \exists \theta : \\
\sup_{\zeta \in \mathcal{K}} ||{f(\zeta) - FNN_\theta (\zeta)}|| < \varepsilon
$$
Onde $\mathcal{K}$ é um conjunto compacto. Portanto, existe uma parametrização capaz de tornar o erro entre a função objetivo e uma FNN arbitrariamente pequeno dentro de um conjunto compacto, porém, encontrar um algoritmo capaz de encontrar essa realização nem sempre é viável.

Ou seja, cada camada realiza uma transformação afim seguida de uma função de ativação, resultando, em geral, em uma transformação não linear.

Diferentes tipos de função de ativação podem ter diferentes resultados sobre o treinamento, os principais e mais comuns de existirem são:
### Função Degrau

![Função Degrau](degrau.svg)

$$
\sigma(a)=
\begin{cases}
0, & a < 0 \\
1, & a \geq 0
\end{cases}
$$
### Função Sigmoide

![Função Sigmoide](sigmoide.svg)

$$
\sigma(a)=\frac{1}{1+e^{-a}}
$$
### Função Tangente Hiperbólica

![Tangente Hiperbólica](tanh.svg)
$$
\sigma(a)=\tanh(a)=\frac{e^a-e^{-a}}{e^a+e^{-a}}
$$
### Função ReLU
![Função ReLU](relu.svg)
$$
\sigma(a)=\max(0,a)
$$

Portando, uma rede neural clássica pode ser entendida como
![Neural Network](neural.svg)

Assim,
O primeiro neurônio calcula:
$$
z_1=\sigma(w_{11}x_1+w_{12}x_2+w_{13}x_3+b_1)
$$

O segundo calcula:
$$
z_2=\sigma(w_{21}x_1+w_{22}x_2+w_{23}x_3+b_2)
$$

O viés é adicionado após a combinação ponderada de todas as entradas e antes da ativação.

Matricialmente:
$$
\begin{bmatrix}
z_1\\z_2
\end{bmatrix}
=
\sigma\left(
\begin{bmatrix}
w_{11}&w_{12}&w_{13}\\
w_{21}&w_{22}&w_{23}
\end{bmatrix}
\begin{bmatrix}
x_1\\x_2\\x_3
\end{bmatrix}
+
\begin{bmatrix}
b_1\\b_2
\end{bmatrix}
\right)
$

Para permitir saídas de redes neurais fora do intervalo da função de ativação das camadas ocultas, pode-se utilizar a última camada sem função de ativação não linear, isto é, com a identidade $\sigma_L(a)=a$.
