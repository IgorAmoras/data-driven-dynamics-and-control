# Feedforward Neural Networks (FNN)

FNNs são a forma mais básica de aprendizado. Propostas em 1900 e bolinha, ela funcionam fazendo uma composição de camadas, cada uma com uma transformação afim seguida de uma função de ativação. O princípio dela é usar de camadas de rede capazes de aproximar a relação entrada-saída do sistema.
FNNs são tidas como aproximadoras universais sob condições adequadas. Na formulação de identificação do mapa de transição de estados apresentada no artigo, necessitamos de ter acesso a todos os estados do sistema para o treinamento supervisionado.
O princípio básico da construção baseaia-se em,
$$
\hat{x}_{i, k+1} = f_\theta(x_{i,k}, u_{i,k}) := FNN(x_{i,k}, u_{i,k})
$$
onde cada FNN consiste em,
$$
FNN(\zeta) := (l_L \circ l_{L-1} \circ ... \circ l_1)(\zeta)
$$
Onde cada camada $l_i$ recebe a saída $z_{i-1}$ de sua antecessora, tendo como valor inicial $z_0=\zeta$. Cada camada é representada como,
$$
z_{i+1}=l_{i+1}(z_i,\theta_{i+1}) = \sigma (W_{i+1}z_i + b_{i+1}), \quad i = 0,...,L - 1
$$
onde $\sigma$ é uma função de ativação, em geral não linear, $W$ são pesos e $b$ são vieses a serem aprendidos, coletados então em $\theta=[\theta^T_1, ... , \theta^T_L]^T$. 
Para uma família de FNNs com ativação adequada e largura suficiente ser tida como aproximadora universal de funções contínuas, estabelecemos formalmente que,
$$
\forall \varepsilon > 0, \quad \exists \theta : \\
\sup_{\zeta \in \mathcal{K}} ||{f(\zeta) - FNN_\theta (\zeta)}|| < \varepsilon
$$
Onde $\mathcal{K}$ é um conjunto compacto. Portanto, existe uma parametrização capaz de tornar o erro entre a função objetivo e uma FNN arbitrariamente pequeno dentro de um conjunto compacto, porém, encontrar um algoritmo capaz de encontrar essa realização nem sempre é viável.

Ou seja, cada camada realiza uma transformação afim seguida de uma função de ativação, resultando, em geral, em uma transformação não linear.

Diferentes tipos de função de ativação podem ter diferentes resultados sobre o treinamento, os principais e mais comuns de existirem são:

| Função Degrau | Função Sigmoide | Função Tangente Hiperbólica | Função ReLU |
|:---:|:---:|:---:|:---:|
| <img src="images/degrau.svg"> | <img src="images/sigmoide.svg"> | <img src="images/tanh.svg"> | <img src="images/relu.svg"> |
| $\displaystyle\sigma(a)=\begin{cases}0,&a<0\\1,&a\geq0\end{cases}$ | $\displaystyle\sigma(a)=\frac{1}{1+e^{-a}}$ | $\displaystyle\sigma(a)=\tanh(a)=\frac{e^a-e^{-a}}{e^a+e^{-a}}$ | $\displaystyle\sigma(a)=\max(0,a)$ |

Portando, uma rede neural clássica pode ser entendida como

</div>

Assim,
O primeiro neurônio calcula:
$$
z_1=\sigma(w_{11}x_1+w_{12}x_2+w_{13}x_3+b_1)
$$

O segundo calcula:
$$
z_2=\sigma(w_{21}x_1+w_{22}x_2+w_{23}x_3+b_2)
$$

O viés é adicionado após a combinação ponderada de todas as entradas e antes da ativação.

Matricialmente:
$$
\begin{bmatrix}
z_1\\z_2
\end{bmatrix}
=
\sigma\left(
\begin{bmatrix}
w_{11}&w_{12}&w_{13}\\
w_{21}&w_{22}&w_{23}
\end{bmatrix}
\begin{bmatrix}
x_1\\x_2\\x_3
\end{bmatrix}
+
\begin{bmatrix}
b_1\\b_2
\end{bmatrix}
\right)
$$

Para permitir saídas de redes neurais fora do intervalo da função de ativação das camadas ocultas, pode-se utilizar a última camada sem função de ativação não linear, isto é, com a identidade $\sigma_L(a)=a$.

Quando várias camadas são consideradas, por exemplo
<div style="text-align:center;">
  <img src="images/neural_2.svg" width="30%">
</div>

Temos que cada camada recebe como entrada sua anterior, matematicamente,

$$
z_0=\zeta=
\begin{bmatrix}
x_1\\x_2\\u
\end{bmatrix}
$$

Primeira camada:
$$
z_1=\sigma_1(W_1z_0+b_1)
$$

Segunda camada:
$$
z_2=\sigma_2(W_2z_1+b_2)
$$

Terceira camada:
$$
z_3=\sigma_3(W_3z_2+b_3)
$$
Agora, se substituirmos recursivamente cada expressão na seguinte, teremos:
$$
\begin{aligned}
f_\theta(\zeta)=\sigma_3\Big(
W_3\sigma_2\big(
&W_2\sigma_1(W_1\zeta+b_1)
&+b_2\big)+b_3
\Big)
\end{aligned}
$$
Assim, recursivamente, definido para um número arbitrário de camadas,

$$
\boxed{
\begin{cases}
z_0=\zeta\\
z_i=\sigma_i(W_i z_{i-1}+b_i),
\quad i=1,\ldots,L\\
f_\theta(\zeta)=z_L
\end{cases}
}
$$
Chegamos na definição formal de uma rede neuronal. 

## Tipos de redes, e escolhas de hiperparâmetros

### Redes largas e profundas
Redes neurais FNNs, tem, em geral, podem ser profundas ou largas,

<div style="text-align:center;">
  <img src="images/tipo_rede.svg" width="50%">
</div>

A escolha da arquitetura de redes é um problema em aberto e existem diferentes resoluções à este problema. Um rede com uma única camada oculta, ams com suficientes parâmetros pode ser capaz de generalizar uma funçional candidato. Enquanto, uma rede profunda, pode representar com menos parâmetros. Isto, entretanto, é analisado de problema a problema.
### Quantidade de neurônios

A quantidade de neurônios determina, juntamente com a profundidade e a ativação, a capacidade de representação da rede. Não necessariamente mais neurônio tem mais expressividade, porém, menos neurônios são capazes de representar uma menor quantidade de informações. Além disso, o custo computacional sobe com a quantidade de neurônios, podendo também, favorecer o sobreajuste.

Para soluções de controle, o ideal é ter a menor rede possível capaz de representar o sistema. 

### Tipo de função de ativação

Diferentemente da quantidade de neurônios, o tipo de função de ativação pode ser baseado nas propriedades matemáticas das função de ativação.
Para aprender, a rede minimiza a perda. $\mathcal{L}(\theta)$ da rede. 

Suponhamos um neurônio,
$$
\hat y=\sigma(wx+b)
$$
para encontrar um $w$ e $b$ capazes de minimizar o erro entre a entrada e a saída da rede, definimos a função de custo tal que,
$$
\mathcal L=\frac{1}{2}(\hat y-y)^2
$$
Para minimizar o erro pela função do gradiente descendente, utilizamos,
$$
w^{(t+1)}=w^{(t)}-\eta
\frac{\partial\mathcal L}{\partial w}
$$
Porém, uma vez que a saída de um neurônio é a entrada do outro, ou seja, eles são dependentes, 
<div style="text-align:center;">
  <img src="images/cadeia.svg" width="33%">
</div>

portanto, o peso influencia a transformação afim, que influencia a ativação, que influencia a saída e, consequentemente, o erro. Assim, matematicamente,
$$
\frac{\partial\mathcal L}{\partial w}=
\frac{\partial\mathcal L}{\partial\hat y}
\frac{\partial\hat y}{\partial a}
\frac{\partial a}{\partial w}
$$
e, individualmente,
$$
\begin{aligned}
\frac{\partial\mathcal L}{\partial\hat y}&=\hat y-y\\
\frac{\partial\hat y}{\partial a}&=\sigma'(a)\\
\frac{\partial a}{\partial w}&=x
\end{aligned}
$$
E enfim, 
$$
\boxed{
\frac{\partial\mathcal L}{\partial w}
=(\hat y-y)\color{#d97706}{\sigma'(a)}x
}
$$

Assim, as derivadas da função de ativação, ponderam como o gradiente descendente irá conseguir aprender os dados.

A função de ativação possui, portanto, dois papéis distintos. Durante a execução, determina a natureza das funções que a rede consegue representar; durante o treinamento, influencia a capacidade de ajustar os pesos e vieses da rede.

### O problema da função degrau

A partir da expressão anterior, tal que $a=wx+b$, podemos escrever a atualização do peso como,

$$
w^{(t+1)}=w^{(t)}-\eta(\hat y-y)\sigma'(a)x
$$

Imaginando uma rede que utiliza a função degrau como função de ativação, temos,

<div style="text-align:center;">
  <img src="images/degrau_derivada.svg" width="65%">
</div>

A derivada é zero em todos os pontos, exceto na origem, onde não existe. Portanto, para $a\neq0$,

$$
\frac{\partial\mathcal L}{\partial w}
=(\hat y-y)\cdot0\cdot x=0
$$

Consequentemente,

$$
w^{(t+1)}=w^{(t)}
$$

Independentemente do erro apresentado pela rede, o peso não será alterado pelo gradiente descendente. O mesmo raciocínio vale para o viés, visto que $\partial\mathcal L/\partial b=(\hat y-y)\sigma'(a)$.

Isto não significa que neurônios com função degrau não possam ser treinados por outros métodos, porém, faz com que esta função de ativação seja inadequada para o treinamento convencional por diferenciação e backpropagation.

### Comparando as funções de ativação durante o treinamento

Suponhamos um neurônio com,

$$
x=1,\quad b=0,\quad y=1
$$

Considerando um peso inicial $w=-1$, uma taxa de aprendizado $\eta=0.5$ e mantendo o viés fixo, podemos analisar o impacto da escolha da ativação em uma iteração de treinamento.

<div style="text-align:center;">
  <img src="images/ativacao_treinamento.svg" width="80%">
</div>

Percebe-se que, para a função degrau e para a ReLU neste ponto de operação negativo, existe erro, porém, a derivada é nula e o peso permanece inalterado. Para a sigmoide e a tangente hiperbólica, o gradiente não é nulo e o algoritmo é capaz de realizar uma correção.

Para a ReLU, se $a>0$, a derivada passa a ser unitária. Já para a sigmoide e a tangente hiperbólica, quando operamos em regiões de saturação, o valor da derivada pode ser muito pequeno, dificultando a atualização dos pesos.

### Impacto da derivada em redes com várias camadas

Imaginando uma rede escalar com três camadas, para simplificar a análise,

$$
\begin{aligned}
a_1&=w_1x+b_1, & z_1&=\sigma_1(a_1)\\
a_2&=w_2z_1+b_2, & z_2&=\sigma_2(a_2)\\
a_3&=w_3z_2+b_3, & z_3&=\sigma_3(a_3)
\end{aligned}
$$

Para uma função de custo $\mathcal L=\frac{1}{2}(z_3-y)^2$, ao calcular a derivada em relação ao peso da primeira camada, pela regra da cadeia, chegamos em,

$$
\frac{\partial\mathcal L}{\partial w_1}
=(z_3-y)\sigma_3'(a_3)w_3
\sigma_2'(a_2)w_2\sigma_1'(a_1)x
$$

Ou seja, existe um produto entre as derivadas das funções de ativação das diferentes camadas e os respectivos pesos. Se considerarmos pesos unitários e derivadas iguais a $0.5$ em todas as camadas, temos,

<div style="text-align:center;">
  <img src="images/gradiente_evanescente.svg" width="60%">
</div>

Conforme retrocedemos pelas camadas, o gradiente pode se tornar cada vez menor. Este problema é conhecido como *vanishing gradient* (gradiente evanescente).

O contrário também pode ocorrer, quando os fatores multiplicativos amplificam o gradiente sucessivamente, causando o *exploding gradient* (gradiente explosivo).

Desta forma, a escolha das funções de ativação está relacionada diretamente à profundidade da rede. Entretanto, as matrizes de pesos também influenciam estes fenômenos, não sendo uma propriedade exclusiva da função de ativação.

### Impacto da função de ativação na capacidade de representação

Até então, analisamos o impacto da derivada das funções de ativação no treinamento. Podemos também analisar o impacto da própria função de ativação na natureza da função representada pela rede.

Suponhamos uma rede com arquitetura,

$$
\mathbb R^1\longrightarrow\mathbb R^2\longrightarrow\mathbb R^1
$$

onde os pesos e vieses são mantidos fixos, tal que,

$$
f(x)=\sigma(x+1)-\sigma(x-1)
$$

<div style="text-align:center;">
  <img src="images/rede_ativacoes.svg" width="55%">
</div>

Alterando exclusivamente a função de ativação, temos,

<div style="text-align:center;">
  <img src="images/comparacao_ativacoes.svg" width="75%">
</div>

A mesma arquitetura, com os mesmos pesos e vieses, é capaz de representar funções de naturezas matemáticas distintas a depender da função de ativação. No exemplo acima, utilizando a identidade, obtemos uma função constante; utilizando a função degrau, uma função descontínua; utilizando a tangente hiperbólica, uma função suave e não linear; e, por fim, utilizando a ReLU, uma função contínua afim por partes.

É importante perceber que, caso todas as funções de ativação de uma FNN sejam a identidade, independentemente da quantidade de camadas, a rede será uma função afim. Por exemplo, para duas camadas,

$$
f_\theta(\zeta)=W_2(W_1\zeta+b_1)+b_2
=W_2W_1\zeta+W_2b_1+b_2
$$

Portanto, não é apenas a quantidade de neurônios ou camadas que determina a capacidade de representação da rede, mas também a natureza das funções de ativação utilizadas.

### Propriedades matemáticas das funções de ativação

| Propriedade | Degrau | Sigmoide | Tanh | ReLU |
|:---|:---:|:---:|:---:|:---:|
| Continuidade | Não | Sim | Sim | Sim |
| Diferenciável em todo domínio | Não | Sim | Sim | Não |
| Suavidade | Descontínua | $C^\infty$ | $C^\infty$ | $C^0$ |
| Imagem | $\{0,1\}$ | $(0,1)$ | $(-1,1)$ | $[0,+\infty)$ |
| Saturação | Ambos os lados | Ambos os lados | Ambos os lados | Lado negativo |
| Afim por partes | Sim (descontínua) | Não | Não | Sim |
| Derivada máxima onde definida | $0$ | $1/4$ | $1$ | $1$ |

As respectivas derivadas são dadas por,

$$
\begin{aligned}
\sigma'_{\mathrm{degrau}}(a)&=0,\quad a\neq 0\\
\sigma'_{\mathrm{sigmoide}}(a)&=\sigma_{\mathrm{sigmoide}}(a)(1-\sigma_{\mathrm{sigmoide}}(a))\\
\sigma'_{\tanh}(a)&=1-\tanh^2(a)\\
\sigma'_{\mathrm{ReLU}}(a)&=\begin{cases}0,&a<0\\1,&a>0\end{cases}
\end{aligned}
$$

A derivada da ReLU e da função degrau não está definida em $a=0$ no sentido clássico.

Uma FNN construída com camadas ocultas ReLU e saída linear é necessariamente uma função contínua afim por partes. Por outro lado, uma FNN construída com tangente hiperbólica e saída linear é uma função infinitamente diferenciável.

Ambas são capazes de aproximar funções contínuas arbitrariamente bem em conjuntos compactos, sob condições adequadas de arquitetura. Entretanto, as estruturas matemáticas utilizadas para realizar esta aproximação são diferentes.

### Impacto na identificação orientada a controle: Duffing

Considerando a dinâmica original do nosso sistema,

$$
\dot{x}=
\begin{bmatrix}
x_2\\
-2x_2-x_1\cos(x_1+x_2)+u
\end{bmatrix}
$$

Temos uma dinâmica suave. Ao identificarmos o mapa de transição utilizando ReLU ou Tanh, teremos duas naturezas distintas de aproximação, como ilustrado para uma função escalar,

<div style="text-align:center;">
  <img src="images/aproximacao_suave_afim.svg" width="65%">
</div>

*Ilustração conceitual de uma aproximação suave e uma aproximação afim por partes de uma função senoidal, não um resultado de treinamento do Duffing.*

Para identificação convencional, ambas podem apresentar erros semelhantes. Para identificação orientada a controle, entretanto, existem outras propriedades de interesse.

Utilizando a tangente hiperbólica, podemos calcular um Jacobiano suave,

$$
J_\theta(x,u)=\frac{\partial f_\theta(x,u)}{\partial x}
$$

Além disso, a Tanh é globalmente Lipschitz com constante $1$, visto que,

$$
0<\sigma'_{\tanh}(a)\leq1
$$

Com a ReLU, o Jacobiano é constante em cada região afim e pode mudar abruptamente nas fronteiras das regiões de ativação, onde pode não existir no sentido clássico.

A estrutura afim por partes abre possibilidades de análise utilizando sistemas politópicos, restrições setoriais, LMIs e ferramentas de otimização mista. A tangente hiperbólica, por sua vez, permite explorar propriedades de suavidade, limites de derivadas e condições de contração. A possibilidade de aplicar tais ferramentas não constitui, por si só, uma garantia de estabilidade do sistema identificado.

Assim, não existe necessariamente uma função de ativação superior, mas existem propriedades matemáticas mais ou menos convenientes para o problema de interesse. Duas redes com praticamente o mesmo erro de identificação podem apresentar propriedades distintas ao tentarmos sintetizar ou certificar um controlador a partir delas.
